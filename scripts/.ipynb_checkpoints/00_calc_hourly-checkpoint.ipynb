```python
dir_path = "/g/data/er8/users/cd3022/Irradiance-comparisons/"
dataset = 'barra-r2'
resolution = 'hourly'
years = []
for year in range(2016, 2025):
    months = []
    for month in range(1, 13):
        date = f'{year}-{month:02d}'
        ds = read_dataset(
                dataset=dataset,
                resolution=resolution,
                date=date
            )
        month_hours = ds['ghi'].groupby("time.hour").mean("time")
        month_hours = month_hours.assign_coords({'month': month})
        months.append(month_hours)
    
    year_ds = xr.concat(months, dim='month')
    year_hours = year_ds.groupby("hour").mean("month")
    year_hours = year_hours.assign_coords({'year': year})
    years.append(year_hours)
    
total = xr.concat(years, dim='year')
total_hourly = total.groupby('hour').mean('year')

total_hourly.to_netcdf(f"{dir_path}hourly_{dataset}.nc")
```
