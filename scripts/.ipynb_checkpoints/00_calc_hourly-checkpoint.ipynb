```python
# BARRA-R2 calculates the hourly irradiance as (start, end] so first time step
# is not included, but the last is (i.e. opposite to normal xarray). To match
# Himawari, need to take the mean of equivalent data

dir_path = "/g/data/er8/users/cd3022/Irradiance-comparisons/"
dataset = 'himawari'

if dataset == 'himawari':
    resolution = 'instant'
else:
    resolution = 'hourly'

year = sys.argv[1]
# --- load all months lazily ---
monthly_datasets = [
    read_dataset(dataset=dataset, resolution=resolution, date=f"{year}-{month:02d}")
    for month in range(1, 13)
]

# --- concat months into one lazy dataset ---
ds_year = xr.concat(monthly_datasets, dim="time")

# --- hourly resample (only for Himawari) ---
da = ds_year.ghi
if dataset == "himawari":
    da = da.resample(time="1H", label="left", closed="right").mean()

# --- compute mean by hour for each month ---
month_hours = (
    da.groupby("time.month")
        .map(lambda m: m.groupby("time.hour").mean("time"))
)

# --- average across months ---
year_hours = month_hours.mean("month")
year_hours = year_hours.assign_coords(year=year)


year_hours.to_netcdf(f"{dir_path}hourly_{dataset}_{year}.nc")
```
